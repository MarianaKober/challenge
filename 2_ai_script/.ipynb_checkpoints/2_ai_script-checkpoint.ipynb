````python
import json
import os
import re
import requests
import psycopg2
import pandas as pd

MANIFEST_PATH = "../loadsmart_dbt/target/manifest.json"
CATALOG_PATH  = "../loadsmart_dbt/target/catalog.json"
MART_SCHEMA   = "analytics"

OLLAMA_URL   = "http://localhost:11434/api/generate"
OLLAMA_MODEL = "llama3.1:8b"  

PG_CONN = dict(
    host="localhost",
    port=5432,
    dbname="loadsmart_dev",
    user="postgres",
    password=os.environ["DBT_PG_PASSWORD"],
)
````

````python
def load_json(path):
    with open(path, encoding="utf-8") as f:
        return json.load(f)

manifest = load_json(MANIFEST_PATH)
catalog = load_json(CATALOG_PATH)
````

````python
def extract_models(manifest, schema=MART_SCHEMA):
    models = {}
    for unique_id, node in manifest["nodes"].items():
        if node.get("resource_type") != "model":
            continue
        if node.get("schema") != schema:
            continue
        columns = {
            col_name: {
                "description": col.get("description", "").strip(),
                "data_type": col.get("data_type"),
            }
            for col_name, col in node.get("columns", {}).items()
        }
        models[node["name"]] = {
            "unique_id": unique_id,
            "description": node.get("description", "").strip(),
            "meta": node.get("meta", {}),
            "relation": f'{node["database"]}.{node["schema"]}.{node["alias"] or node["name"]}',
            "columns": columns,
        }
    return models

models = extract_models(manifest)
print(list(models.keys()))
````

````python
def enrich_with_catalog(models, catalog):
    catalog_nodes = catalog["nodes"]
    for name, model in models.items():
        node = catalog_nodes.get(model["unique_id"])
        if not node:
            continue
        for col_name, col_meta in node.get("columns", {}).items():
            if col_name in model["columns"]:
                model["columns"][col_name]["data_type"] = col_meta.get("type")

enrich_with_catalog(models, catalog)
````

````python
def extract_relationships(manifest):
    rels = []
    for unique_id, node in manifest["nodes"].items():
        if node.get("resource_type") != "test":
            continue
        meta = node.get("test_metadata", {})
        if meta.get("name") != "relationships":
            continue
        kwargs = meta.get("kwargs", {})
        args = kwargs.get("arguments", kwargs)
        depends_on = node.get("depends_on", {}).get("nodes", [])
        rels.append({
            "from_column": args.get("column_name") or kwargs.get("column_name"),
            "to_model_ref": args.get("to"),
            "to_field": args.get("field"),
            "from_model_unique_id": depends_on[0] if depends_on else None,
        })
    return rels

def clean_ref(ref_str):
    m = re.search(r"ref\('([^']+)'\)", ref_str or "")
    return m.group(1) if m else ref_str

relationships = extract_relationships(manifest)
````

````python
def build_schema_context(models, relationships):
    lines = [
        "### SCHEMA CONTEXT",
        "Use ONLY the tables, columns, and relationships defined below.",
        ""
    ]
    
    # Pre-build lookup for relationships
    id_to_name = {v["unique_id"]: k for k, v in models.items()}
    
    for name, model in models.items():
        relation = model.get("relation", name)
        lines.append(f"### Table: `{relation}`")
        
        if model.get("description"):
            lines.append(f"Description: {model['description'].strip()}")

        if model.get("ai_hint"):
            lines.append(f"AI Hint: {model['ai_hint'].strip()}")
            
        if model.get("meta"):
            meta_str = ", ".join(f"{k}={v}" for k, v in model["meta"].items())
            lines.append(f"Metadata: [{meta_str}]")
            
        lines.append("Columns:")
        for col_name, col in model.get("columns", {}).items():
            dtype = f" ({col['data_type']})" if col.get("data_type") else ""
            desc = f" — {col['description'].strip()}" if col.get("description") else ""
            lines.append(f"- `{col_name}`{dtype}{desc}")
            
        lines.append("")  # Blank line for clean section separation

    if relationships:
        lines.append("### RELATIONSHIPS")
        for r in relationships:
            from_name = id_to_name.get(r.get("from_model_unique_id"), "?")
            to_name = clean_ref(r.get("to_model_ref", ""))
            from_col = r.get("from_column", "?")
            to_field = r.get("to_field", "?")
            lines.append(f"- `{from_name}`.`{from_col}` -> `{to_name}`.`{to_field}`")

    return "\n".join(lines)

schema_context = build_schema_context(models, relationships)
print(schema_context)
````

````python
def generate_sql(question, schema_context, max_d=None, model=OLLAMA_MODEL):


    prompt = f"""### Task
Generate a PostgreSQL query to answer the following question:
`{question}`

### Rules
- Output ONLY the final SQL query. No comments, no explanations, no pseudocode, no markdown fences.
- The output must start directly with the word SELECT.
- NEVER include any columns that are no explicit asked in the question
- When calculating an average or median, always exclude zero values (e.g., by using NULLIF(column_name, 0) or filtering zeros out in the WHERE clause)
- Use ONLY the fully-qualified table names exactly as given in the schema (e.g. database.schema.table).
- Every column reference must carry its table name (table.column), and every table referenced must appear in FROM or JOIN. A column that belongs to a dimension table exists only there, so JOIN that table on its key before using the column.
- from, into, to, must genereate filters in the where clause
- most, lest, higher, lower must generate filters for top dim required by measure
- When writing SQL queries that contain a `GROUP BY` clause: 1. Ensure EVERY non-aggregated column in the `SELECT`, `HAVING`, or `ORDER BY` clauses is explicitly included in the `GROUP BY` list.
- Only JOIN a table if a column you need is not available in the tables already used. If you JOIN, use the correct key and never reference a column from a table you haven't joined.
- ALWAYS write every column as table.column. Every table that owns a column you use MUST appear in FROM or JOIN. A dimension column exists only in its dim_ table, never in fact_load, so JOIN it with its key.
- If the question mentions a JOIN is needed, you MUST include the JOIN clause with the correct key. Never reference a column from a table you haven't joined.
- COMPARISONS ("compare", "vs", "versus", "differ", "difference between", "how does A compare to B"): return ONE ROW PER compared value. GROUP BY the column that holds the compared values and SELECT that column next to the aggregate. If the question names specific values (for example TX and CA), use WHERE column IN ('A','B') together with GROUP BY column. If the values are the categories of a column (for example intrastate and interstate in lane_type), GROUP BY that column with no WHERE on it. Never repeat the same aggregate twice under different aliases. Never filter to only one of the compared values.
- If a column represents a metric or requires row-level calculation inside an aggregated query, you MUST wrap it in an aggregate function (e.g., `SUM()`, `AVG()`, `MAX()`, `MIN()`).
- Never calculate unaggregated ratios or expressions across rows alongside a `GROUP BY`. Either aggregate the components first or include the columns in the `GROUP BY` list if grouping by individual row values.
- "per load" means one row of the load table, so "average X per load" is simply AVG(NULLIF(X, 0)). Do NOT divide by anything.
- For questions like "which X has the most/highest/top", GROUP BY only X, ORDER BY the aggregate DESC, and add LIMIT 1 (or LIMIT N if the question says "top N").
- Never divide one column by another unless the question explicitly asks for a ratio or says "per <unit>". Only then use SUM(a) / NULLIF(SUM(b), 0).
- Only use columns that the question's words refer to. Do not add extra columns or calculations that the question does not ask for.
- Every filter in the WHERE clause MUST map directly to a specific keyword or entity in the user's question.
- Do NOT add dates as a default partition or optimization strategy.
- NEVER include date filters UNLESS the question explicitly mentions a specific year, date range, or time period
- CRITICAL: Only add a WHERE condition for a column if the question explicitly refers to that concept. Do NOT add filters on tracking flags.

### Database Schema
The query will run on a database with the following schema. Only use these tables and columns.
{schema_context}

### SQL Query
```sql
"""

    response = requests.post(
        OLLAMA_URL,
        json={
            "model": model,
            "prompt": prompt,
            "stream": False,
            "options": {"temperature": 0},
        },
        timeout=600,
    )
    response.raise_for_status()
    raw = response.json()["response"]

    match = re.search(r"select.*?;", raw, re.IGNORECASE | re.DOTALL)
    if match:
        sql = match.group(0)
    else:
        match = re.search(r"select.*", raw, re.IGNORECASE | re.DOTALL)
        sql = (match.group(0).strip() + ";") if match else raw.strip()

    return sql.strip()
````

````python
teste_sql = generate_sql("Quantos loads foram entregues no total?", schema_context)
print(teste_sql)
````

````python
def execute_sql(sql, limit=200):
    conn = psycopg2.connect(**PG_CONN)
    try:
        with conn.cursor() as cur:
            cur.execute(f"SET search_path TO {MART_SCHEMA}, public;")
        df = pd.read_sql(sql, conn)
        return df.head(limit)
    finally:
        conn.close()

execute_sql(teste_sql)
````

````python
def answer_question(question):
    try:
        sql = generate_sql(question, schema_context)
    except requests.exceptions.ReadTimeout:
        return {"question": question, "sql": None, "result": None, "error": "Ollama timeout ao gerar SQL"}
    try:
        df = execute_sql(sql)
        return {"question": question, "sql": sql, "result": df, "error": None}
    except Exception as e:
        return {"question": question, "sql": sql, "result": None, "error": str(e)}
questions = [
    "How many loads were delivered in the last full month available in the data?"
]

results = [answer_question(q) for q in questions]
for r in results:
    print("—" * 60)
    print("Q:", r["question"])
    print("SQL:", r["sql"])
    print("Erro:", r["error"])
````

````python
results_table = pd.DataFrame([
    {
        "question": r["question"],
        "generated_sql": r["sql"],
        "answer": r["result"].to_dict(orient="records") if r["result"] is not None else None,
        "error": r["error"],
    }
    for r in results
])
results_table.to_csv("results_table.csv", index=False)
results_table
````
